```python
test_heat = torch.full([1,15,128,128], -5)

    
test_heat[0, 5, 62, 73] = 2
test_heat[0,8, 72, 97] = 3
test_heat[0, 1, 24, 54] = 4

peak_map = decoder.decoder(test_heat)
flat_peak_score, flat_peak_postion = decoder.top_k_peaks(peak_map, K=3)
dims = decoder.make_dims(flat_peak_score, flat_peak_postion)

print(dims)

```

```python
x = torch.zeros(1, 2, 128, 128)
y = torch.zeros(1, 2, 128, 128)

x[0, 0, dims[1], dims[2]] = torch.tensor([5, 6, 7], dtype= torch.float32)

x[0, 1, dims[1], dims[2]] = torch.tensor([5, 6, 7], dtype= torch.float32)

y[0, 0, dims[1], dims[2]] = torch.tensor([0.4, 0.6, 0.8], dtype= torch.float32)

y[0, 1, dims[1], dims[2]] = torch.tensor([0.3, 0.6, 0.7], dtype= torch.float32)



results = decoder.make_box(offset =y , size = x, row = dims[1], col = dims[2], score = flat_peak_score, class_num = dims[0])

print(results)
```
